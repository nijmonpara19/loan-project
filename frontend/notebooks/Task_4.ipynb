```python
X = df.drop(["LoanID", "Default"], axis=1)
y = df["Default"]
```

```python
X.select_dtypes(include="object").columns
```

```python
## for Employment Type
X["EmploymentType"] = X["EmploymentType"].map({
    "Unemployed": 0,
    "Part-time": 1,
    "Full-time": 2,
    "Self-employed": 3
})
```

```python
X["MaritalStatus"] = X["MaritalStatus"].map({
    "Single": 0,
    "Married": 1,
    "Divorced": 2
})
```

```python
X["HasMortgage"] = X["HasMortgage"].map({
    "No": 0,
    "Yes": 1
})
```

```python
X["HasDependents"] = X["HasDependents"].map({
    "No": 0,
    "Yes": 1
})
```

```python
## for Loan Purpose
X["LoanPurpose"] = X["LoanPurpose"].map({
    "Auto": 0,
    "Business": 1,
    "Education": 2,
    "Home": 3,
    "Other": 4
})
```

```python
## for Has Co-Signer
X["HasCoSigner"] = X["HasCoSigner"].map({
    "No": 0,
    "Yes": 1
})
```

```python
## for education cols
X["Education"] = X["Education"].map({
    "High School": 0,
    "Bachelor's": 1,
    "Master's": 2,
    "PhD": 3
})
```
